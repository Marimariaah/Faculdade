# Análise de Dados - Clientes de Cartão de Crédito
## Exercícios de Análise de Características Financeiras

Este notebook contém soluções para os 7 exercícios de análise exploratória de dados sobre clientes de cartão de crédito.

```python
# Importar bibliotecas necessárias
import pandas as pd  #São bibliotecas Para manipulação de grandes conjuntos de dados
import numpy as np   #São bibliotecas Para manipulação de grandes conjuntos de dados
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar estilo dos plots
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

# Caminho do arquivo Excel
arquivo_excel = r'C:\Dev\Faculdade\Periodo 03\CIÊNCIA DOS DADOS PARA TOMADAS DE DECISÕES\Trabalho\Atividade 02\default_of_credit_card_clients__courseware_version_1_21_19 (1).xls'

# Carregar dados
df = pd.read_excel(arquivo_excel)

print("Dataset carregado com sucesso!")
print(f"Forma do dataset: {df.shape}")
print(f"\nPrimeiras linhas do dataset:")
print(df.head())
```

## Exercício 1: Criar Listas com Nomes para as Características Financeiras

Vamos criar listas organizadas com os nomes das características financeiras do dataset.

```python
# Exercício 1: Criar listas com nomes para características financeiras

# Mostrar todas as colunas do dataset
print("Colunas do dataset:")
print(df.columns.tolist())

# Criar listas organizadas por categorias de características financeiras

# Características de Valor da Fatura (Bill Amounts) - geralmente estão em colunas separadas por mês
bill_amounts = ['BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6']

# Características de Valor do Pagamento (Payment Amounts) - geralmente estão em colunas separadas por mês
payment_amounts = ['PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3', 'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']

# Características de Status de Pagamento (Payment Status) - repayment status
payment_status = ['PAY_1', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']

# Características Demográficas
demographic_features = ['AGE', 'SEX', 'EDUCATION', 'MARITAL', 'LIMIT_BAL']

print("\n✓ Listas de características criadas com sucesso!\n")
print("Lista de Valores da Fatura (6 meses):")
print(bill_amounts)
print("\nLista de Valores de Pagamento (6 meses):")
print(payment_amounts)
print("\nLista de Status de Pagamento (6 meses):")
print(payment_status)
print("\nLista de Características Demográficas:")
print(demographic_features)
```

## Exercício 2: Examinar Sínteses Estatísticas dos Valores da Fatura

Use .describe() para examinar as características de valor da fatura e refletir sobre os resultados.

```python
# Exercício 2: Usar .describe() para valores da fatura

print("RESUMO ESTATÍSTICO DOS VALORES DE FATURA (6 MESES):")
print("=" * 80)

bill_statistics = df[bill_amounts].describe()
print(bill_statistics)

print("\n" + "=" * 80)
print("\nREFLEXÃO SOBRE OS RESULTADOS:")
print("-" * 80)
print("""
1. Contagem (count): Verificar se há dados faltantes em relação ao total de registros.
2. Média (mean): A média dos valores de fatura representa a tendência central.
3. Desvio Padrão (std): Mede a variabilidade dos valores. Um desvio alto indica grande 
   dispersão nos saldos de fatura entre clientes.
4. Mínimo (min): Os valores mínimos (frequentemente 0) indicam clientes com saldo zero.
5. Percentis (25%, 50%, 75%): 
   - 25% (Q1): 25% dos clientes tem fatura menor que este valor
   - 50% (Q2/Mediana): Valor central da distribuição
   - 75% (Q3): 75% dos clientes tem fatura menor que este valor
6. Máximo (max): Os valores máximos podem indicar limite de crédito ou clientes de alto gasto.

INTERPRETAÇÃO: Faz sentido ver uma grande variação nos valores de fatura, refletindo
a diversidade de padrões de consumo e limites de crédito entre os clientes.
""")
```

## Exercício 3: Visualizar Valores da Fatura com Grid 2x3 de Histogramas

Criar uma grade de 2 por 3 plotagens de histogramas para os 6 meses de valores de fatura, usando 20 bins.

```python
# Exercício 3: Grid 2x3 de histogramas para valores de fatura

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Distribuição dos Valores de Fatura (6 Meses)', fontsize=16, fontweight='bold', y=1.00)

# Flatten the axes array para facilitar a iteração
axes = axes.flatten()

# Criar um histograma para cada coluna de valor de fatura
for idx, bill_col in enumerate(bill_amounts):
    ax = axes[idx]
    df[bill_col].hist(bins=20, ax=ax, edgecolor='black', alpha=0.7, color='steelblue')
    ax.set_title(f'{bill_col}', fontweight='bold')
    ax.set_xlabel('Valor (R$)')
    ax.set_ylabel('Frequência')
    ax.grid(True, alpha=0.3)
    
    # Adicionar estatísticas no título
    media = df[bill_col].mean()
    mediana = df[bill_col].median()
    ax.text(0.95, 0.95, f'Média: R${media:,.0f}\nMediana: R${mediana:,.0f}', 
            transform=ax.transAxes, verticalalignment='top', horizontalalignment='right',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5), fontsize=9)

plt.tight_layout()
plt.show()

print("ANÁLISE DOS HISTOGRAMAS DE VALORES DE FATURA:")
print("-" * 80)
print("""
Observações:
- A maioria dos valores de fatura apresenta distribuição assimétrica (skewed).
- Há concentração de valores em faixas menores, com cauda longa para valores maiores.
- Alguns meses podem mostrar padrões sazonais ou diferentes distribuições.
- A presença de picos em certos valores pode indicar pontos de crítica ou limites.
""")
```

## Exercício 4: Obter Resumo describe() para Valores de Pagamento

Use .describe() para examinar as características de valor do pagamento.

```python
# Exercício 4: Usar .describe() para valores de pagamento

print("RESUMO ESTATÍSTICO DOS VALORES DE PAGAMENTO (6 MESES):")
print("=" * 80)

payment_statistics = df[payment_amounts].describe()
print(payment_statistics)

print("\n" + "=" * 80)
print("\nREFLEXÃO SOBRE OS RESULTADOS:")
print("-" * 80)
print("""
1. Contagem (count): Verificar se há dados faltantes.
2. Média (mean): Valor médio de pagamento mensal. Comparar com média de faturas.
3. Desvio Padrão (std): Variabilidade nos pagamentos entre clientes.
4. Mínimo (min): Frequentemente 0, indicando clientes que não pagaram.
5. Percentis (25%, 50%, 75%): Distribuição dos padrões de pagamento.
6. Máximo (max): Pagamentos máximos (possivelmente para liquidar saldos).

INTERPRETAÇÃO: Faz sentido observar uma grande quantidade de zeros e uma distribuição
assimétrica, pois nem todos os clientes fazem pagamentos todo mês ou alguns fazem
pagamentos parciais. A diferença entre faturas e pagamentos é crucial para 
análise de risco de crédito.
""")

# Comparison between bill and payment
print("\nCOMPARAÇÃO ENTRE FATURA E PAGAMENTO:")
print("-" * 80)
bill_mean = df[bill_amounts].mean().mean()
payment_mean = df[payment_amounts].mean().mean()
print(f"Média de Faturas: R$ {bill_mean:,.2f}")
print(f"Média de Pagamentos: R$ {payment_mean:,.2f}")
print(f"Diferença Média: R$ {bill_mean - payment_mean:,.2f}")
print(f"\nIsso indica que em média, os clientes pagam {(payment_mean/bill_mean)*100:.1f}% de suas faturas.")
```

## Exercício 5: Plotar Histogramas de Pagamento com Rótulos Rotacionados

Criar histogramas para valores de pagamento similar ao Exercício 3, mas com rotação dos rótulos do eixo x.

```python
# Exercício 5: Grid 2x3 de histogramas para valores de pagamento com rotação

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Distribuição dos Valores de Pagamento (6 Meses)', fontsize=16, fontweight='bold', y=1.00)

# Flatten the axes array
axes = axes.flatten()

# Criar um histograma para cada coluna de valor de pagamento
for idx, pay_col in enumerate(payment_amounts):
    ax = axes[idx]
    df[pay_col].hist(bins=20, ax=ax, edgecolor='black', alpha=0.7, color='coral')
    ax.set_title(f'{pay_col}', fontweight='bold')
    ax.set_xlabel('Valor (R$)')
    ax.set_ylabel('Frequência')
    ax.grid(True, alpha=0.3)
    
    # Rotacionar rótulos do eixo x
    ax.tick_params(axis='x', rotation=45)
    
    # Adicionar estatísticas no título
    media = df[pay_col].mean()
    mediana = df[pay_col].median()
    ax.text(0.95, 0.95, f'Média: R${media:,.0f}\nMediana: R${mediana:,.0f}', 
            transform=ax.transAxes, verticalalignment='top', horizontalalignment='right',
            bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5), fontsize=9)

plt.tight_layout()
plt.show()

print("ANÁLISE DOS HISTOGRAMAS DE VALORES DE PAGAMENTO:")
print("-" * 80)
print("""
Observações:
1. PICOS EM ZERO: Há um pico muito grande em zero, indicando clientes que não fizeram 
   pagamentos em alguns meses (possível inadimplência ou períodos sem consumo).
   
2. DISTRIBUIÇÃO ASSIMÉTRICA: Os pagamentos que não são zero também mostram distribuição
   com cauda longa, com maioria em valores menores e poucos pagamentos muito altos.
   
3. DIFERENÇA COM FATURAS: Comparado aos histogramas de fatura, os pagamentos mostram
   muito mais concentração em zero e distribuição mais limitada, sugerindo que muitos
   clientes não pagam a fatura inteira.
   
4. COMPORTAMENTO POR MÊS: Pode haver variações entre meses, refletindo padrões sazonais
   ou diferentes comportamentos de pagamento ao longo do período.
""")
```

## Exercício 6: Usar Máscara Booleana para Contar Pagamentos Iguais a Zero

Use uma máscara booleana para verificar quantos valores de pagamento são exatamente iguais a 0.

```python
# Exercício 6: Usar máscara booleana para contar pagamentos iguais a zero

print("ANÁLISE DE PAGAMENTOS IGUAIS A ZERO:")
print("=" * 80)

# Criar uma máscara booleana para cada coluna de pagamento
zero_counts = {}
total_records = len(df)

for pay_col in payment_amounts:
    # Máscara booleana: valores iguais a zero
    mask_zeros = df[pay_col] == 0
    count_zeros = mask_zeros.sum()
    percentage = (count_zeros / total_records) * 100
    
    zero_counts[pay_col] = {
        'count': count_zeros,
        'percentage': percentage
    }
    
    print(f"\n{pay_col}:")
    print(f"  Pagamentos = 0: {count_zeros:,} ({percentage:.2f}%)")
    print(f"  Pagamentos > 0: {total_records - count_zeros:,} ({100 - percentage:.2f}%)")

# Resumo geral
print("\n" + "=" * 80)
print("RESUMO GERAL DE PAGAMENTOS ZERO:")
print("-" * 80)

all_zeros = []
for pay_col in payment_amounts:
    all_zeros.append(zero_counts[pay_col]['count'])

print(f"Total de instâncias onde pagamento = 0: {sum(all_zeros):,}")
print(f"Total de registros: {total_records * len(payment_amounts):,}")
print(f"Percentual médio de pagamentos zero: {np.mean([v['percentage'] for v in zero_counts.values()]):.2f}%")

print("\nINTERPRETAÇÃO:")
print("""
O alto número de pagamentos iguais a zero faz TOTAL SENTIDO dado que:
1. Muitos clientes podem não estar utilizando o cartão em alguns meses.
2. Clientes em mora não estão fazendo novos pagamentos.
3. Alguns clientes podem estar em período de carência.
4. O PICO VISÍVEL NO HISTOGRAMA no valor zero corresponde exatamente 
   a esses registros detectados pela máscara booleana.

A máscara booleana confirma a observação visual do histograma anterior!
""")
```

## Exercício 7: Aplicar Transformação Logarítmica aos Pagamentos Não-Zero

Ignorar pagamentos iguais a zero usando máscara booleana, aplicar transformação logarítmica (log10) 
com .apply() e plotar histogramas dos dados transformados.

```python
# Exercício 7: Transformação logarítmica de pagamentos não-zero

# Criar figura para comparação: original vs transformado
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Distribuição dos Pagamentos: Original vs Log10 Transformado', 
             fontsize=16, fontweight='bold', y=0.995)

axes_flat = axes.flatten()

for idx, pay_col in enumerate(payment_amounts):
    # ===== Linha 1: Dados originais (apenas não-zero) =====
    ax_original = axes[0, idx]
    
    # Máscara booleana para filtrar pagamentos > 0
    mask_nonzero = df[pay_col] > 0
    payments_nonzero = df[pay_col][mask_nonzero]
    
    payments_nonzero.hist(bins=20, ax=ax_original, edgecolor='black', alpha=0.7, color='lightblue')
    ax_original.set_title(f'{pay_col} (Valores > 0)', fontweight='bold')
    ax_original.set_xlabel('Valor (R$)')
    ax_original.set_ylabel('Frequência')
    ax_original.grid(True, alpha=0.3)
    
    # ===== Linha 2: Dados transformados com log10 =====
    ax_log = axes[1, idx]
    
    # Aplicar transformação logarítmica usando .apply() e np.log10()
    payments_log = payments_nonzero.apply(np.log10)
    
    payments_log.hist(bins=20, ax=ax_log, edgecolor='black', alpha=0.7, color='lightgreen')
    ax_log.set_title(f'{pay_col} (Log10 Transformado)', fontweight='bold')
    ax_log.set_xlabel('Log10(Valor)')
    ax_log.set_ylabel('Frequência')
    ax_log.grid(True, alpha=0.3)
    
    # Adicionar estatísticas
    info_text = f'N={len(payments_nonzero)}\nZeros: {(df[pay_col] == 0).sum()}'
    ax_original.text(0.95, 0.95, info_text, transform=ax_original.transAxes,
                    verticalalignment='top', horizontalalignment='right',
                    bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5), fontsize=8)

plt.tight_layout()
plt.show()

print("ANÁLISE DA TRANSFORMAÇÃO LOGARÍTMICA:")
print("=" * 80)
print("""
MOTIVAÇÃO PARA TRANSFORMAÇÃO LOGARÍTMICA:
Os dados originais de pagamento têm distribuição altamente assimétrica (skewed)
com muitos valores pequenos e alguns valores muito grandes. A transformação 
logarítmica ajuda a:
- Normalizar a distribuição
- Reduzir o impacto de outliers
- Tornar a variância mais uniforme
- Facilitar análises estatísticas que assumem distribuição normal

INTERPRETAÇÃO DOS RESULTADOS:
1. ANTES (Original):
   - Distribuição com cauda longa para a direita
   - Muito concentrada em valores baixos
   - Difícil de visualizar toda a amplitude

2. DEPOIS (Log10):
   - Distribuição muito mais aproximada à normal
   - Valores espalhados de forma mais uniforme
   - Mais fácil identificar padrões
   - Melhor para modelos de regressão e análises estatísticas

QUANTIDADE DE DADOS:
""")

for pay_col in payment_amounts:
    mask = df[pay_col] > 0
    total = len(df)
    nonzero = mask.sum()
    zeros = total - nonzero
    print(f"{pay_col}: {nonzero:,} pagamentos não-zero ({nonzero/total*100:.1f}%) | {zeros:,} zeros ({zeros/total*100:.1f}%)")

print("""
CONCLUSÃO:
A transformação logarítmica dos pagamentos não-zero revela uma distribuição muito
mais próxima à normal, o que é benéfico para análises estatísticas e modelagem
de dados financeiros. Os zeros foram ignorados conforme solicitado, pois log(0)
é indefinido e esses representam uma categoria separada (clientes que não pagaram).
""")
```
